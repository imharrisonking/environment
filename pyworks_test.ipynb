# Test Notebook for pyworks.nvim

This notebook tests the integration of pyworks.nvim with image display.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', linewidth=2, label='sin(x)')
plt.title('Test Plot for pyworks.nvim')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Test another plot type
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Subplot 1
ax1.scatter(np.random.randn(50), np.random.randn(50), alpha=0.6)
ax1.set_title('Random Scatter Plot')

# Subplot 2
ax2.hist(np.random.randn(1000), bins=30, alpha=0.7)
ax2.set_title('Random Histogram')

plt.tight_layout()
plt.show()
```
